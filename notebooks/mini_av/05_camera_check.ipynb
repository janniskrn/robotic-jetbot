# 05 Camera check

Live view of the robot's camera, to place the robot. Below it: the spots of the resolution test (2026-11-11).

1. Run the first code cell. Move the robot until the live view looks about like the spot picture.
2. Run the last cell (**stop the camera**) before you tell Claude "spot N ready": the picture script needs the camera.
3. The picture script restarts Jupyter, so this notebook stops then. Run it again from the top for the next spot.

The motors are not used here.

```python
import ipywidgets.widgets as widgets
import traitlets
from IPython.display import display
from jetbot import Camera, bgr8_to_jpeg

SPOTS = '/workspace/usb/images/situation_check_2026-11-11/spots.jpg'  # the three spots, side by side

camera = Camera.instance()
live = widgets.Image(format='jpeg', width=448, height=448)  # the 224 x 224 camera picture, shown twice as large
camera_link = traitlets.dlink((camera, 'value'), (live, 'value'), transform=bgr8_to_jpeg)
with open(SPOTS, 'rb') as file:
    spots = widgets.Image(value=file.read(), format='jpeg', width=960)

display(widgets.Label('live camera'), live, widgets.Label('spots to match'), spots)
```

Stop the camera (run before saying "spot N ready").

```python
camera_link.unlink()
camera.stop()
print('camera stopped: Claude can take the pictures now')
```
